### 2.ndarray's attributes
- [shape]------shape:number of rows & number of columns <br>
- [ndim]-------dimension:0 or 1 or 2 <br>
- [size]-------number:the total count of the elements <br>
- [dtype]------type:the type of an element in an array(integer,float,string) <br>
Also some high level attributes: T , itemsize , nbytes , flags<br>
- *.T (transpose):* It returns the transpose of the array, swapping rows and columns.
- *.itemsize:* It shows how many bytes one single element takes up.
- *.nbytes:* It gives the total bytes occupied by all elements in the array.
- *.flags:* It stores memory layout flags of the ndarray, such as whether the data is stored contiguously.

```python
#Quick Check for "shape" and "size" and "dtype"
arr1 = np.array(1)
print('The 0-dimensional array\'s shape:',arr1.shape)
print(f'There are(is) {arr1.size} element(s) in this array')
print(f'The type of the elements in this array is {arr1.dtype}')
print()

arr2 = np.array(['he','she','it'])
print('The 1-dimensional array\'s shape:',arr2.shape)
print(f'There are(is) {arr2.size} element(s) in this array')
print(f'The type of the elements in this array is {arr2.dtype}')
print()

arr3 = np.array([[1.3,2.2,3.1],[4.6,5.5,6.4]])
print('The 2-dimensional array\'s shape:',arr3.shape)
print(f'There are(is) {arr3.size} element(s) in this array')
print(f'The type of the elements in this array is {arr3.dtype}')
print()
```

# ②How to create a ndarray

| Purpose                | Methods                                                                 |
| ---------------------- | ----------------------------------------------------------------------- |
| Basic construction    | `np.array()` (create an ndarray, 创建数组) <br> `np.copy()` (make a copy of array, 复制数组) |
| Fill with predefined shape | `np.zeros()` (fill with 0s, 全0数组) <br> `np.ones()` (fill with 1s, 全1数组) <br> `np.empty()` (uninitialized array, 空数组，不初始化) <br> `np.full()` (fill with a fixed value, 用指定值填充) |
| Generate from numerical range | `np.arange()` (range like Python range, 按步长生成序列) <br> `np.linspace()` (evenly spaced numbers, 等分采样) <br> `np.logspace()` (logarithmically spaced numbers, 对数等分) |
| Special matrix creation | `np.eye()` (identity matrix, 单位矩阵) <br> `np.diag()` (diagonal matrix / extract diagonal, 对角矩阵 / 提取对角线) |
| Random array generation | `np.random.rand()` (uniform [0,1), 均匀分布随机数) <br> `np.random.randn()` (standard normal, 标准正态分布) <br> `np.random.randint()` (random integers, 随机整数) |
| Advanced construction | `np.array()` <br> `np.loadtxt()` (load data from text file, 读取文本文件数据) <br> `np.fromfunction()` (build array from function, 根据函数生成数组) |


### 1.Basic construction

```python
#array
list1 = [4,5,6]
arr = np.array(list1)
print(arr)

    #additionally, we can froce the change of elements' type:
arr = np.array(list1,dtype=str)
print(arr)
```

```python
#copy
arr_new = np.copy(arr)
print(arr_new)
#Attention:arr_new and arr are DIFFERENT arraies
```

### 2.Fill with predefined shape

```python
#zeros
arr = np.zeros((2,3))
print(arr)

print('or')

arrr = np.zeros((3,),dtype=int)
print(arrr)
```

```python
#ones
arr = np.ones((2,3))
print(arr)
    #same as zeros above
```

```python
#empty
arr = np.empty((4,3))
print(arr)   #random
```

```python
#full
arr = np.full((3,4),66)#66 is chosen by yourself
print(arr)
```

```python
#xxx_like
arr = np.full((4,5),12)
print(arr)
arr0 = np.zeros_like(arr) #'xxx_like' means create an array which has the same shape as the liked one
print(arr0)
```

### 3.Generate from numerical range

```python
#arange
arr = np.arange(1,10,1) #start,end,step
print(arr)
```

```python
#linspace
arr = np.linspace(1,10,5) #start,end,num(ivide into equal parts)
print(arr)
```

```python
#logspace
arr = np.logspace(0,4,3,base=2)  #start,end,num,calculate(log).
#if we dont write 'base=?',it will calculate log10 automatically.
arr10 = np.logspace(0,4,3)
print(arr)
print(arr10)
```

### 4.Special matrix creation

```python
#eye
arr = np.eye(3,dtype=int)
print(arr)

arrr = np.eye(4,6,dtype=int)
print(arrr)
```

```python
#diag
arr = np.diag([1,4,2,7,8])
print(arr)
```

### 5.Random array generation 

```python
#random.rand
arr = np.random.rand(3,4)
print(arr)
#random number[float] belongs to [0,1),shaped 3*4
```

```python
#random.uniform
arr = np.random.uniform(4,5,(4,3))
print(arr)
#random unmber[float] belongs to [4,5),shaped 4*3
```

```python
#random.randint
arr = np.random.randint(2,10,(4,2))
print(arr)
# random numbers[integer] belongs to [2,10),shaped 4*2
```

```python
#random.randn
arr = np.random.randn(2,3)
print(arr)
#random numbers following a normal distribution
```

```python
#random.seed
    #**Random seed**
    #A random seed is an initial value for the pseudorandom number generator.
    #If you set the same seed, you will get exactly the same sequence of random numbers every time. 
    #It makes random results reproducible.
np.random.seed(66)
arr = np.random.randint(1,10,(4,2))
print(arr)
```
