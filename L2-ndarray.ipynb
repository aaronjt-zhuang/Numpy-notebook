# Data types in ndarray

| Data Type（数据类型） | Description（说明） |
| ---- | ---- |
| bool | Boolean type（布尔类型） |

| int16, uint16 | Signed / unsigned 16-bit (2-byte) integer（有符号、无符号的 16 位（2 字节）整型） |

| float32 | Single-precision floating-point（单精度浮点型） |
| float64 | Double-precision floating-point（双精度浮点型） |
| complex64 | Complex number represented by two 32-bit floats（用两个 32 位浮点数表示的复数） |
| complex128 | Complex number represented by two 64-bit floats（用两个 64 位浮点数表示的复数） |

# Index and Slice

| Index / Slice Type（索引/切片类型） | Description / Usage（描述/用法） |
| ---- | ---- |
| Basic indexing（基本索引） | Access elements directly by integer index. Index starts at 0. <br> 通过整数索引直接访问元素。索引从 0 开始。 |
| Row / column slicing（行/列切片） | Use colon `:` slice syntax to select a subset of rows or columns. <br> 使用冒号切片语法选择行或列的子集。 |
| Continuous slicing（连续切片） | Slice from the start index to the end index with a given step. <br> 从起始索引到结束索引按步长切片。 |
| Using the slice function（使用 slice 函数） | Define slicing rules via `slice(start, stop, step)`. <br> 通过 `slice(start, stop, step)` 定义切片规则。 |
| Boolean indexing（布尔索引） | Filter elements that satisfy a Boolean condition. Supports logical operators `&`, `\|`. <br> 通过布尔条件筛选满足条件的元素。支持逻辑运算符 `&`、`\|`。 |


```python
#1-dimensional array's index and slice
arr = np.random.randint(1,100,20)
print(arr)
print(arr[0])
print(arr[2:6])
print(arr[slice(2,6)])#the same output as above
print(arr[slice(2,6,2)])#start,end,step
print(arr[arr>50]) #boolean index
```

```python
#2-dimensional array's index and slice
np.random.seed(20)
arr = np.random.randint(1,100,(5,5))
print(arr)
print()
#index:if we want the '41' in the middle of this array
print(arr[2,2]) #[x-1,y-1], for the index starts with 0
print()
#what about slice?
print(arr[:,:])#this means select All
print()
print(arr[0,:])#this means select the 1st row wholly
print()
print(arr[1,1:3])#this means select the 2ed to 3rd in row2
print()
print(arr[1:4,1:4])#this means select the inner 3*3 array
print()
#what about boolean?
print(arr[arr>50])#all the number more than 50
print()
print(arr[2][arr[2]>50]) #the numbers more than 50 in row3
print()
print(arr[:,2][arr[:,2]>50])#the numbers more than 50 in column3

```
